```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

print(df.shape)
df.head()
```

```python
df.info()

df.describe()
```

```python
# Only Sunday bills
sunday = df[df["day"] == "Sun"]
print(len(sunday))
sunday.head()


big_dinners = df[(df["total_bill"] > 30) & (df["time"] == "Dinner")]
print(len(big_dinners))

print(len(df[df["smoker"] == "Yes"]))
print(len(df[df["size"] > 4]))
print(len(df[(df["day"] == "Sat") & (df["tip"] > 5)]))
```

```python
df.groupby("day")["total_bill"].sum()
```

```python
df.groupby("smoker")["tip"].mean()
```

```python
df.groupby("time")["total_bill"].mean()
```

```python
df["tip_pct"] = df["tip"] / df["total_bill"] * 100

df.head()
```

```python
df["per_person"] = df["total_bill"]/df["size"]

df.head()
```

```python
tip_pct = df.groupby("day")["tip_pct"].mean()

print(tip_pct)
```

```python
Tip_pct_for_smokers = df.groupby("smoker")["tip_pct"].mean()
print(Tip_pct_for_smokers)
```

```python
df.sort_values("tip_pct", ascending=False).head()
```
